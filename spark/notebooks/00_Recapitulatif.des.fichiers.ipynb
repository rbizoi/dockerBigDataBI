```python
import sqlalchemy, pandas as pd
from sqlalchemy import text
from sqlalchemy.engine import URL


url = URL.create(
    drivername="postgresql+psycopg2",
    username="formation",
    password="formation",
    host="postgres-source",
    port=5432,
    database="formation",
)

engine = sqlalchemy.create_engine( 
    url, #url = "postgresql://formation:formation@postgres-source:5432/formation"
    pool_pre_ping=True,
)

with engine.connect() as connexion:
    print(connexion.execute(text("SELECT current_database(), current_user")).fetchone())

connection = engine.connect()    
```

```python
requete  = "select * from customers"
pd.read_sql_query(requete, connection)
```

```python
requete  = """
            select 'orders     ' table, count(*) enregistrements      from orders       
            union all
            select 'customers  ' , count(*) from customers    
            union all
            select 'sellers    ' , count(*) from sellers      
            union all
            select 'items      ' , count(*) from items        
            union all
            select 'payments   ' , count(*) from payments     
            union all
            select 'reviews    ' , count(*) from reviews      
            union all
            select 'reviews_p  ' , count(*) from reviews_p    
            union all
            select 'products   ' , count(*) from products     
            union all
            select 'geolocation' , count(*) from geolocation"""
pd.read_sql_query(requete, connection)
```
