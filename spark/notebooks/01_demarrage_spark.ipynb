```python
import os, warnings

warnings.filterwarnings(action="ignore")
```

```python
from pyspark.sql import SparkSession, functions as F

spark = (SparkSession.builder
         .appName("JupyterLab-Training")
         .getOrCreate())

print("Master :", spark.sparkContext.master)
print("Application :", spark.sparkContext.applicationId)
print("Python :", os.sys.version.split()[0])
print("Spark :", spark.version)
```

```python
resultat = (spark.range(1, 1001, numPartitions=4)
            .groupBy((F.col("id") % 2).alias("parite"))
            .count()
            .orderBy("parite"))
resultat.show()
```

Après l'exécution, l'application est visible dans Spark Master puis dans Spark History Server.
